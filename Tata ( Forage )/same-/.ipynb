```python
import pandas as pd

def compare_excel(file1, file2):
    print("checking 1...")
    df1 = pd.read_excel(file1)
    print("checking 2...")
    df2 = pd.read_excel(file2)

    if df1.equals(df2):
        print("Both Excel files are SAME ✔️")
    else:
        print("Excel files are DIFFERENT ❌")

compare_excel("Online Retail.xlsx", "Online Retail Data Set.xlsx")
```
